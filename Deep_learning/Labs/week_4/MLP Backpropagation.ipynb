```python
import numpy as np
from sklearn import metrics
np.set_printoptions(precision=6)
```

```python
def sigmoid(v):
    return 1/(1+np.exp(-v))
```

```python
def sigmoid_derivative(v):
    return sigmoid(v)*(1-sigmoid(v))
```

# Backpropagation with a single instance

```python
x = np.array([0.5,0.1])
y = 0.9
l=0.5
```

**Forward Pass**

```python
w1 = np.array([[0.15, 0.2],[0.24, 0.3]])
b1 = np.array([1,1])
w2 = np.array([[-0.4],[0.5]])
b2 = 1

#Forward Pass
inL1 = x@w1 + b1
print('Input to layer 1: ',inL1)
outL1 = sigmoid(inL1)
print('Output of layer 1: ',outL1)
inL2 = outL1@w2 + b2
print('Input to layer 2: ', inL2)
outL2 = sigmoid(inL2)
print('Output of layer 2: ',outL2)
```

**Error Calculation**

```python
w1 = np.array([[0.15, 0.2],[0.24, 0.3]])
b1 = np.array([1,1])
w2 = np.array([[-0.4],[0.5]])
b2 = 1

#Forward Pass
inL1 = x@w1 + b1
outL1 = sigmoid(inL1)
inL2 = outL1@w2 + b2
outL2 = sigmoid(inL2)

#Error calulation
e = 0.5*np.power(y-outL2, 2)
print('Error: ', e)
```

**Backpropagation Layer 2**

```python
w1 = np.array([[0.15, 0.2],[0.24, 0.3]])
b1 = np.array([1,1])
w2 = np.array([[-0.4],[0.5]])
b2 = 1

#Forward Pass
inL1 = x@w1 + b1
outL1 = sigmoid(inL1)
inL2 = outL1@w2 + b2
outL2 = sigmoid(inL2)

#Error calulation
e = 0.5*np.power(y-outL2, 2)

#Backpropagation layer 2
dEdOutL2 = outL2 - y
print('dEdOutL2: ', dEdOutL2)
dOutL2dInL2 = sigmoid_derivative(inL2)
print('dOutL2dInL2: ', dOutL2dInL2)
dInL2dW2 = outL1
print('dInL2dW2: ', dInL2dW2)

dEdW2 = outL1.reshape([len(outL1),1])*dEdOutL2*dOutL2dInL2
print('dEdW2: ',dEdW2)

dEdB2 = dEdOutL2*dOutL2dInL2
print('dEdB2: ',dEdB2)
```

```python
w1 = np.array([[0.15, 0.2],[0.24, 0.3]])
b1 = np.array([1,1])
w2 = np.array([[-0.4],[0.5]])
b2 = 1

#Forward Pass
inL1 = x@w1 + b1
outL1 = sigmoid(inL1)
inL2 = outL1@w2 + b2
outL2 = sigmoid(inL2)

#Error calulation
e = 0.5*np.power(y-outL2, 2)

#Backpropagation layer 2
dEdW2 = outL1.reshape([len(outL1),1])*sigmoid_derivative(inL2)*(outL2 - y)
dEdB2 = sigmoid_derivative(inL2)*(outL2 - y)

#Backpropagation layer 1
dInL2dOutL1 = w2
print('dInL2dOutL1: ', dInL2dOutL1)
dOutL1dInL1 = sigmoid_derivative(inL1)
print('dOutL1dInL1: ', dOutL1dInL1)
dInL1dW1 = x
print('dInL1dW1: ', dInL1dW1)

dEdW1 = dInL1dW1.reshape([len(dInL1dW1),1])@(dEdOutL2*dOutL2dInL2*dInL2dOutL1.T)*dOutL1dInL1
print('dEdW1: ', dEdW1)

dEdB1 = (dEdOutL2*dOutL2dInL2*dInL2dOutL1.T)*dOutL1dInL1
print('dEdB1: ',dEdB1)
```

```python
#Read data from a text file
from sklearn import preprocessing
data = np.loadtxt("simpleregression.txt")

# split into inputs and outputs
x, y = data[:, :-1], data[:, -1]
y=y.reshape(len(y),1)

#scaling the input data
#X = preprocessing.MinMaxScaler().fit_transform(X)

print ("X.shape:", x.shape)
print ("y.shape:", y.shape)
```

```python
np.random.seed(42)
w1 = np.random.uniform(-1,1,[len(x[0]),2]) #weights of the first layer
b1 = np.zeros([1,2]) #bias of the first layer
w2 = np.random.uniform(-1,1,[2,1]) #weights of the first layer
b2 = 0 #bias of the first layer

#Forward Pass
inL1 = x@w1 + b1
outL1 = sigmoid(inL1)
inL2 = outL1@w2 + b2
outL2 = sigmoid(inL2)

#Error calulation
e = 0.5*np.power(y-outL2, 2)
print('Error before: ', e.mean())

#Backpropagation layer 2
dEdW2 = (1/x.shape[0])*outL1.T@(sigmoid_derivative(inL2)*(outL2 - y))
dEdB2 = (1/x.shape[0])*sigmoid_derivative(inL2)*(outL2 - y)

#Backpropagation layer 2
dEdW1 = (1/x.shape[0])*x.T@(((sigmoid_derivative(inL2)*(outL2 - y))@(w2.T))*sigmoid_derivative(inL1))
dEdB1 = (1/x.shape[0])*((sigmoid_derivative(inL2)*(outL2 - y)*(w2.T))*sigmoid_derivative(inL1))

#updating parameters
w1 = w1 - l*dEdW1
b1 = b1 - l*dEdB1
w2 = w2 - l*dEdW2
b2 = b2 - l*dEdB2

#Forward Pass
inL1 = x@w1 + b1
outL1 = sigmoid(inL1)
inL2 = outL1@w2 + b2
outL2 = sigmoid(inL2)

#Error calulation
e = 0.5*np.power(y-outL2, 2)
print('Error after: ',e.mean())
```
